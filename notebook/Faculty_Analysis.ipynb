```python
# =========================
# Faculty Survey Analysis 
# =========================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
import re

# ------------------------
# Load data
# ------------------------
df = pd.read_csv("Faculty Survey_anonymized.csv", dtype=str)  # read as strings first to avoid mixed dtypes

print("Shape:", df.shape)
print("Preview:")
display(df.head(2))

# ------------------------
# Drop admin/metadata columns 
# ------------------------
meta_cols = [
    "StartDate", "EndDate", "Progress", "Duration (in seconds)",
    "Finished", "RecordedDate", "Start Date", "End Date", "Recorded Date"
]
df = df.drop(columns=[c for c in meta_cols if c in df.columns], errors="ignore")

# ------------------------
# Rename key columns 
# ------------------------
rename_map = {
    "Q1.3": "Gender", "Q1.4": "Race",
    "Q1.5": "Years Teaching", "Q1.6": "Programs Taught",
    "Q1.7": "Undergraduate Programs", "Q1.8": "Graduate Programs"
}
df.rename(columns={k:v for k,v in rename_map.items() if k in df.columns}, inplace=True)

# ------------------------
# Normalize Likert text → numbers (case-insensitive)
# ------------------------
likert_map_lower = {
    "not at all important": 1,
    "slightly important": 2,
    "moderately important": 3,
    "very important": 4,
    "extremely important": 5
}

def maybe_map_likert(series: pd.Series) -> pd.Series:
    # Work on a lowercased, stripped view
    s = series.astype(str).str.strip().str.lower()
    # If at least half of non-null entries are known Likert labels, map them
    mask_known = s.isin(likert_map_lower.keys())
    if mask_known.mean() >= 0.5:
        out = s.map(likert_map_lower)
        return out  # numeric ints 1-5 (floats if NaN present)
    else:
        return series  # leave unchanged

# Apply to all object-like columns
for col in df.columns:
    df[col] = maybe_map_likert(df[col])

# ------------------------
# Coerce rank/numeric columns safely
#    (fixes the "str + int" bug by unifying dtypes)
# ------------------------
converted_numeric_cols = []

def try_numeric_col(series: pd.Series, threshold=0.8) -> pd.Series:
    # attempt to coerce; if most values convert, keep numeric
    coerced = pd.to_numeric(series, errors="coerce")
    share_numeric = coerced.notna().mean()
    return coerced if share_numeric >= threshold else series

for col in df.columns:
    before_dtype = df[col].dtype
    new_series = try_numeric_col(df[col])
    if not new_series.equals(df[col]):
        df[col] = new_series
        converted_numeric_cols.append(col)

# Also force any obvious ranking blocks (e.g., Q2.3_*) to numeric
rank_blocks = ["Q2.3_", "Q2.4_", "Q4.2_", "Q4.3_", "Q4.4_", "Q4.5_", "Q5.2_"]
rank_cols = [c for c in df.columns if any(c.startswith(rb) for rb in rank_blocks)]
for c in rank_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")
    if c not in converted_numeric_cols:
        converted_numeric_cols.append(c)

print("\nConverted to numeric:", converted_numeric_cols)

# ------------------------
#  Quick demographics 
# ------------------------
if "Gender" in df.columns:
    print("\nGender distribution:")
    print(df["Gender"].value_counts(dropna=False))

if "Race" in df.columns:
    print("\nRace distribution:")
    print(df["Race"].value_counts(dropna=False))

if "Years Teaching" in df.columns:
    df["Years Teaching"] = pd.to_numeric(df["Years Teaching"], errors="coerce")
    print("\nYears Teaching (describe):")
    print(df["Years Teaching"].describe())

# ------------------------
# Likert summary (averages of 1–5 after mapping)
# ------------------------
numeric_cols = df.select_dtypes(include=[np.number]).columns
likert_means = df[numeric_cols].mean().sort_values(ascending=False)

print("\nNumeric columns averaged (includes Likert + ranks):")
display(likert_means.head(20))

# If you want ONLY the Likert-style columns (1–5), keep those with max<=5 and min>=1
likert_only = df[numeric_cols].loc[:, (df[numeric_cols].max() <= 5) & (df[numeric_cols].min() >= 1)]
likert_only_means = likert_only.mean().sort_values(ascending=False)
print("\nLikert-only means (1–5):")
display(likert_only_means.head(20))

# ------------------------
# Ranking analysis (lower = higher priority)
# ------------------------
if rank_cols:
    rank_df = df[rank_cols]
    rank_means = rank_df.mean().sort_values()  # lower mean rank -> higher priority
    print("\nAverage rankings (lower = higher priority):")
    display(rank_means)

# ------------------------
# Open-ended responses (example columns)
# ------------------------
# Try common text columns if present; add yours here
text_candidates = [c for c in df.columns if c.endswith("_TEXT") or c in ["Q3.8","Q5.3","Q5.4","Q5.5","Q5.6","Q30_TEXT","Q37_TEXT","Q40_TEXT","Q46_TEXT","Q47_TEXT","Q48_TEXT"]]
text_candidates = [c for c in text_candidates if c in df.columns]

def top_words(series, n=20):
    text = " ".join(series.dropna().astype(str))
    tokens = re.findall(r"\w+", text.lower())
    # simple stoplist (expand as needed)
    stop = set(["the","and","to","of","in","for","a","is","on","that","with","as","it","be","are","this","at","we"])
    tokens = [t for t in tokens if t not in stop and not t.isdigit()]
    return Counter(tokens).most_common(n)

for col in text_candidates:
    tw = top_words(df[col], n=20)
    if tw:
        print(f"\nTop words for {col}:")
        print(tw)

# ------------------------
# Visuals (single-plot, default colors)
# ------------------------
plt.figure()
if "Gender" in df.columns:
    df["Gender"].value_counts(dropna=False).plot(kind="bar", title="Faculty Gender Distribution")
    plt.xlabel("Gender"); plt.ylabel("Count")
    plt.show()

plt.figure()
if "Race" in df.columns:
    df["Race"].value_counts(dropna=False).plot(kind="bar", title="Faculty Race/Ethnicity Distribution")
    plt.xlabel("Race/Ethnicity"); plt.ylabel("Count")
    plt.show()

plt.figure()
if not likert_only_means.empty:
    likert_only_means.plot(kind="barh", title="Average Importance (Likert 1–5)")
    plt.xlabel("Average Score")
    plt.ylabel("Question")
    plt.gca().invert_yaxis()  # highest at top
    plt.show()
```
